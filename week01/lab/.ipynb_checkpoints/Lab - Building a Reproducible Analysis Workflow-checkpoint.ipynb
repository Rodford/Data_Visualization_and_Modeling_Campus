# Part 1: Florida park visitation


## Background

The National Park Service (NPS) publishes estimates of recreational visits to its units. These estimates help parks understand patterns in public use and plan staffing, maintenance, transportation, and visitor services.

This course file contains monthly estimates for five Florida NPS units from 2015 through 2025:

- Big Cypress National Preserve
- Biscayne National Park
- Canaveral National Seashore
- Dry Tortugas National Park
- Everglades National Park

A recreational visit is a visit, not necessarily one unique person. A person who visits several times may be counted several times.


| Feature | Description |
| --- | --- |
| `park_code` | Four-letter NPS unit code |
| `park` | Name of the NPS unit |
| `year` | Calendar year of the observation |
| `month_number` | Month represented as an integer from 1 through 12 |
| `month` | Month name |
| `recreation_visits` | Estimated recreational visits during that park-month |


## Step 1: Question

**After the decline in July 2020, when did Canaveral National Seashore's July recreational visitation first return to or exceed its average July visitation from 2015–2019?**

The 2015–2019 mean provides a pre-pandemic comparison point. In this question, “recovery” means the first July from 2020 onward with visitation at or above that baseline. This definition does not require every later July to remain above the baseline.


## Step 2: Data

Import Pandas using the alias `pd`. Read `data/nps_florida_park_visitation_2015_2025.csv` into a DataFrame named `park_visits`, then display its first five rows.


```python
import pandas as pd

park_visits = pd.read_csv('data/nps_florida_park_visitation_2015_2025.csv')
park_visits.head()
```

Display the last five rows.


```python
park_visits.tail()
```

Display information about the DataFrame and calculate the number of missing observations in each feature.


```python
park_visits.info()

print("\nMissing values:")
park_visits.isna().sum()
```

Your answer: Each row represents the total number of recreation visits to one specific Florida National Park during a single month and year.

**Are any observations missing from the six features?**

Your answer: No, there are zero missing observations. The data summary shows exactly 660 non-null entries for all 660 rows, and the missing value count is exactly 0 for every feature.

## Step 3: Operation

Create one Boolean condition that identifies Canaveral National Seashore and another that identifies July.


```python
canaveral_condition = park_visits['park'] == 'Canaveral National Seashore'
july_condition = park_visits['month'] == 'July'
```

Combine the two conditions with `&`. Use `.loc` to select the `year` and `recreation_visits` features, make a copy named `canaveral_july`, and display it.


```python
canaveral_july = park_visits.loc[canaveral_condition & july_condition, ['year', 'recreation_visits']].copy()
display(canaveral_july)
```

Create `pre_pandemic_condition`, which is `True` for years from 2015 through 2019. Use it with `.loc` to calculate the mean of `recreation_visits`, and store the result as `pre_pandemic_july_mean`.

Display `pre_pandemic_july_mean`.


```python
pre_pandemic_condition = (canaveral_july['year'] >= 2015) & (canaveral_july['year'] <= 2019)
pre_pandemic_july_mean = canaveral_july.loc[pre_pandemic_condition, 'recreation_visits'].mean()
display(pre_pandemic_july_mean)
```

Create a condition that identifies rows from 2020 onward whose `recreation_visits` are greater than or equal to `pre_pandemic_july_mean`. Use `.loc` to create `recovery_years`, then display it in ascending year order.


```python
recovery_condition = (canaveral_july['year'] >= 2020) & (canaveral_july['recreation_visits'] >= pre_pandemic_july_mean)
recovery_years = canaveral_july.loc[recovery_condition].sort_values(by='year', ascending=True)
display(recovery_years)
```

## Step 4: Check

Display the number of rows in `canaveral_july`, its smallest year, and its greatest year.


```python
print("Number of rows:", len(canaveral_july))
print("Smallest year:", canaveral_july['year'].min())
print("Greatest year:", canaveral_july['year'].max())
```

**Do these results confirm that the filtered data contains one July observation for every year from 2015 through 2025?**

Your answer: Yes. The span from 2015 to 2025 covers exactly 11 inclusive years, which perfectly matches the 11 total rows in the DataFrame.

Count the observations used to calculate `pre_pandemic_july_mean`.

**Does the baseline use the expected five years?**

Your answer: Yes. The output shows exactly 5 observations, which corresponds to the five pre-pandemic years (2015, 2016, 2017, 2018, and 2019).

Display the first row of `recovery_years`.

**What was the 2015–2019 mean, and which year first met or exceeded it after the July 2020 decline? Report that year's July visitation.**

## Step 6: Conclusion


**Write one sentence that answers the research question using the evidence.**

Your answer: July visitation at Canaveral National Seashore first recovered to its pre-pandemic average levels in 2021.

## Step 7: Limitation


**Does first exceeding the baseline mean that every later July remained above it? Use the data to explain.**

Your answer: No. The data shows that the year 2023 is absent from the recovery_years table, meaning July 2023 visitation (173,296 visits) dipped back below the pre-pandemic baseline of 173,971.2 before exceeding it again in 2024.

Your answer: This analysis only tracks raw visitation numbers over time, establishing correlation rather than causation. Proving the pandemic directly caused the decline would require external data on specific variables like mandated park closures, travel restrictions, or weather events during July 2020.

## Step 1: Question

Formulate one research question that can be answered with your dataset. The question must require a calculation or comparison, not merely looking up one value.


**What is your research question?**

Your answer: Did games in the Action genre or the Sports genre generate higher total global sales?

## Step 2: Data

Read your CSV into a clearly named DataFrame. Display the first and last five rows, information about the DataFrame, and the missing-value count for every feature.


```python
import pandas as pd

# Read the CSV into a DataFrame
vg_sales = pd.read_csv('data/vgsales.csv')

# Display the first five rows
display(vg_sales.head())

# Display the last five rows
display(vg_sales.tail())

# Display DataFrame information
vg_sales.info()

# Display the missing-value count for every feature
print("\nMissing values:")
display(vg_sales.isna().sum())
```

**Which observations and features will you use to answer your question?**

Your answer: I will use observations where the game's genre is either "Action" or "Sports." The specific features I will use are Genre to filter the data, and Global_Sales to calculate the total sales for each category.

## Step 3: Operation

Write the code needed to prepare, filter, group, summarize, or sort the data. Store the final evidence in a clearly named object and display it.


```python
# Filter the data for only Action and Sports genres
target_genres = vg_sales['Genre'].isin(['Action', 'Sports'])
filtered_sales = vg_sales.loc[target_genres]

# Group by genre, calculate the total global sales, and sort descending
genre_sales_comparison = filtered_sales.groupby('Genre')['Global_Sales'].sum().sort_values(ascending=False)

# Display the final evidence object
display(genre_sales_comparison)
```

**Explain how your operations connect the selected data to your research question.**

Your answer: First, I filtered the dataset to isolate only the observations belonging to the "Action" and "Sports" genres. Next, I grouped these isolated records by their genre and calculated the total sum of their Global_Sales. Finally, I sorted the results in descending order to definitively show which of the two genres produced the higher total sales volume, which directly answers the research question.

## Step 4: Check

Write at least one check that could reveal an error in your analysis, then display its result.


```python
calculated_sum = round(genre_sales_comparison.sum(), 2)
expected_total = round(filtered_sales['Global_Sales'].sum(), 2)

print("Sum of grouped genre sales:", calculated_sum)
print("Total global sales of filtered dataset:", expected_total)
print("Do they match?", calculated_sum == expected_total)
```

**What did you check, and did the result meet your expectation?**

Your answer: I checked whether the rounded sum of the aggregated sales for the individual Action and Sports categories exactly equaled the total global sales sum of the filtered dataset. The result met my expectation by returning True, confirming that the grouped data completely accounts for all sales records in the filtered dataset without any loss or duplication.

## Step 5: Evidence


**Which specific value or comparison in your output answers the question?**

Your answer: The grouped sales comparison shows that Action games generated 1,751.18 million in global sales, while Sports games generated 1,330.93 million, proving that Action games achieved higher total global sales.

## Step 6: Conclusion


**What conclusion is supported by the evidence?**

Your answer: Video games in the Action genre generated a higher total volume of global sales than those in the Sports genre within this dataset.

## Step 7: Limitation


**What should a reader avoid concluding from your analysis, and why?**

Your answer: A reader should avoid concluding that Action games are consistently more profitable or popular on an individual title basis, because this comparison only measures aggregate total sales across all titles combined and does not account for the sheer volume or release frequency of games published in each genre.
