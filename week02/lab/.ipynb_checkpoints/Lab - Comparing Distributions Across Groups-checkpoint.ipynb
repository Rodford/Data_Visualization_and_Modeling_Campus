# Guided analysis

## Step 1: Question

> **How did the daily number of reported street-condition requests differ across New York City's five boroughs in 2025?**


## Step 2: Data

### Import the libraries

Import Pandas, Matplotlib's `pyplot`, and Seaborn using their conventional aliases.


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

### Read the data

Read `data/nyc_311_street_conditions_2025.csv` into a DataFrame named `requests`. Use `parse_dates=["created_date"]` so Pandas converts the CSV's text timestamps into datetime values while loading the file.


```python
requests = pd.read_csv("data/nyc_311_street_conditions_2025.csv", parse_dates=["created_date"])
```

### Preview the data

Display the first and last five rows.


```python
display(requests.head())
display(requests.tail())
```

### Inspect the DataFrame

Display its stored data types and missing-value counts.


```python
print("Data Types:")
print(requests.dtypes)
print("\nMissing Values:")
print(requests.isnull().sum())
```

**Complete the table using the data dictionary and your inspection results.** Classify each feature as categorical, quantitative, or temporal.

**What does one row represent?**

Your answer: One row represents a single 311 service request created in 2025 reporting a street condition in New York City. It does not necessarily represent one unique physical street problem, as multiple residents could report the same issue.

### Predict


**Which borough do you predict will have the greatest typical number of reported street-condition requests per day? What informed your prediction?**

Your answer: I predict Queens will have the greatest typical number of requests. Queens is geographically the largest with the most street mileage, which naturally creates a larger surface area for potholes and street defects to occur and be reported.

## Step 3: Operation


### Make the borough names human-readable

The source stores borough names in all capital letters. **Title case** capitalizes the first letter of each word and writes the remaining letters in lowercase. For example, `STATEN ISLAND` becomes `Staten Island`.

Use `.str.title()` to convert every value in `requests["borough"]` to title case and save the result back into the same feature.


```python
requests["borough"] = requests["borough"].str.title()
```

**What roles do `.str` and `.title()` play in this statement?**

Your answer: The .str accessor allows Pandas to apply string manipulation methods to every individual value in a Series. The .title() method is the specific string operation that capitalizes the first letter of every word and makes all subsequent letters lowercase.

### Create a calendar date

Create `requests["date"]` with `requests["created_date"].dt.normalize()`. The `.dt` accessor opens Pandas' datetime tools, and `.normalize()` sets the time in every value to midnight while preserving its calendar date.


```python
requests["date"] = requests["created_date"].dt.normalize()
```

### Count requests for each borough and date

Create a DataFrame named `daily_requests` with this sequence:

1. group `requests` by `date` and `borough`;
2. use `.size()` to count rows in every group; and
3. use `.reset_index(name="request_count")` to return a DataFrame whose count column is named `request_count`.

Display the first five rows.


```python
daily_requests = requests.groupby(["date", "borough"]).size().reset_index(name="request_count")
display(daily_requests.head())
```

**What does one row of `daily_requests` represent? How is that different from one row of `requests`?**

Your answer: One row in daily_requests represents the total tally of street condition requests reported on one specific calendar date for one specific borough.

**Why did we remove the time before grouping and counting the requests?** As a hint, try repeating the calculation with `created_date` in place of `date` and inspect the result.

Your answer: If we kept the exact time, almost every group would contain exactly 1 request because it is incredibly rare for two 311 calls to be logged at the exact same millisecond. Grouping by precise timestamps prevents us from calculating meaningful daily totals.

### Set the borough order

Create `borough_order` containing the five borough names in alphabetical order.


```python
borough_order = ["Bronx", "Brooklyn", "Manhattan", "Queens", "Staten Island"]
```

### Calculate the mean daily request count

Create `borough_summary` by grouping `daily_requests` by `borough`, selecting `request_count`, calculating the mean, and using `.reindex(borough_order)`.


```python
borough_summary = daily_requests.groupby("borough")["request_count"].mean().reindex(borough_order)
display(borough_summary)
```

**Why is comparing only these means incomplete?**

Your answer: The mean only provides a single measure of center. It completely obscures the shape of the data distribution, the spread, how often the counts overlap between boroughs, and whether there are extreme outliers.

## Step 4: Check


### Check borough-date coverage

Count the number of rows for each borough in `daily_requests`, then use `.reindex(borough_order)`.


```python
display(daily_requests.groupby("borough").size().reindex(borough_order))
```

**Does every borough have a count for all 365 dates?**

Your answer: Yes, each of the five boroughs has 365 rows.

**What would it mean if one or more boroughs had a count of fewer than 365 rows?**

Your answer: It would mean there was at least one day out of the year where a borough had zero street-condition requests reported, preventing a group from forming for that date. Alternatively, it could mean the underlying data file was missing records for specific dates entirely.

### Check the calculated data

Display missing-value counts for `date`, `borough`, and `request_count`. Then use `.describe()` to summarize `request_count` within each borough and put the boroughs in `borough_order`.


```python
display(daily_requests[["date", "borough", "request_count"]].isnull().sum())
display(daily_requests.groupby("borough")["request_count"].describe().reindex(borough_order))
```

**Are any calculated values missing or impossible?**

Your answer: There are no missing values. There are also no impossible values; the minimum daily counts are safely above zero, and maximums are realistic bounds for a large city system.

**Would a zero necessarily be impossible in another 311 dataset?**

Your answer: No. In a dataset for a smaller city, a less common complaint type, or a smaller geographic grouping, having zero reports on a given day would be entirely possible and expected.

## Step 5: Evidence


### Create overlapping histograms

Use `sns.histplot()` with:

- `data=daily_requests`;
- `x="request_count"`;
- `hue="borough"` and `hue_order=borough_order`;
- `element="step"`;
- layered histograms;
- `alpha=0.45`; and
- the `"deep"` palette.

Capitalize the legend title and add a question-focused title and descriptive axis labels. Label the y-axis `Number of days`.


```python
sns.histplot(
    data=daily_requests,
    x="request_count",
    hue="borough",
    hue_order=borough_order,
    element="step",
    alpha=0.45,
    palette="deep"
)
plt.title("How do daily 311 street-condition requests compare across NYC boroughs?")
plt.xlabel("Number of daily requests")
plt.ylabel("Number of days")

# Capitalize the legend title
plt.legend(title="Borough", labels=borough_order[::-1]) # Reverse to match default stack order
plt.show()
```

**What does the histogram reveal, and what makes the boroughs easy or difficult to compare?**

Your answer: It reveals that Queens and Brooklyn have distributions shifted much further to the right (higher daily counts), while Staten Island and the Bronx cluster to the left. However, the heavy visual overlap of five different colored polygons makes it very difficult to read individual peaks, spreads, or precise median points.

### Create a box plot

Put `request_count` on the x-axis and `borough` on the y-axis. Use `borough_order`, the named color `"mediumseagreen"`, and descriptive labels.


```python
sns.boxplot(
    data=daily_requests,
    x="request_count",
    y="borough",
    order=borough_order,
    color="mediumseagreen"
)
plt.title("Distribution of Daily 311 Street-Condition Requests by Borough")
plt.xlabel("Number of daily requests")
plt.ylabel("Borough")
plt.show()
```

**What differences among boroughs are easier to see in the box plot than in the histogram?**

Your answer: The box plot clearly isolates the median line and the interquartile range for each borough without any overlapping visual clutter. It also clearly identifies exact extreme outliers as individual dots.

### Create a violin plot

Use the same x feature, y feature, and order. Change only the plot color to `"mediumpurple"` in addition to the required data mappings and order.


```python
sns.violinplot(
    data=daily_requests,
    x="request_count",
    y="borough",
    order=borough_order,
    color="mediumpurple"
)
plt.title("Distribution of Daily 311 Street-Condition Requests by Borough")
plt.xlabel("Number of daily requests")
plt.ylabel("Borough")
plt.show()
```

**What does the violin plot add to the comparison?**

Your answer: The violin plot adds a smoothed estimate of the data's density shape. Unlike the box plot, which assumes a uniform box, the violin plot reveals where exactly the values are most heavily clustered, showing if the data is heavily skewed or bimodal.

**Which of the three figures most clearly compares daily request counts among boroughs? Explain your choice.**

Your answer: The box plot is the most effective. Because our primary goal is to compare centers and spreads across distinct categories, the box plot aligns these metrics clearly on a single axis without the visual noise of the overlapping histogram or the abstract density curves of the violin plot.

## Step 6: Conclusion


**What conclusion about borough differences is supported by the 2025 data?**

Your answer: Queens consistently experiences the highest daily volume of reported street conditions by a significant margin, followed by Brooklyn. In contrast, the Bronx and Staten Island reliably report the lowest daily request volumes, with Manhattan sitting near the middle.

## Step 7: Limitation


**Why do these request counts not directly measure the number of street problems or borough performance?**

Your answer: 311 data purely measures the volume of complaints. A higher count in a borough might be driven by a larger population, more widespread awareness of the 311 app, or greater civic willingness to report issues, rather than strictly reflecting poorer street quality or slower government repair times.

# Your own analysis

Develop a new research question using the 311 data. Your question must compare a **quantitative feature across categorical groups**. You will probably need to calculate the quantitative feature from the original request records.

Possible grouping features include `descriptor`, month, or day of the week. You may use another meaningful categorical feature that you can create and explain.


## Step 1: Question


**Write a research question that compares a quantitative feature across categorical groups.**

Your answer: How does the daily volume of NYC street-condition requests differ between the days of the week?

## Step 2: Data


**Which original features will you use, and what will each row of your calculated DataFrame represent?**

Your answer: I will use the date feature created earlier to extract the day of the week, and request_count. Each row of the grouped DataFrame will represent the total number of street-condition requests reported city-wide on a single calendar day, categorized by its day of the week.

## Step 3: Operation

Prepare the grouping feature, calculate the quantitative feature, and save the resulting DataFrame with a meaningful name.


```python
# Create a total citywide daily count
citywide_daily = daily_requests.groupby("date")["request_count"].sum().reset_index()

# Extract the day of the week
citywide_daily["day_of_week"] = citywide_daily["date"].dt.day_name()

# Order the days properly
day_order = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]
citywide_daily["day_of_week"] = pd.Categorical(citywide_daily["day_of_week"], categories=day_order, ordered=True)

display(citywide_daily.head())
```

**Briefly explain how your code produced the quantitative feature and comparison groups.**

Your answer: I grouped the existing daily_requests table by date to sum all boroughs together into a citywide daily request_count. Then, I used the datetime .dt.day_name() method to extract the day of the week from the date, converting it into an ordered categorical feature so the days plot in chronological order (Monday to Sunday).

## Step 4: Check

Check the calculated data for missing or impossible values, confirm that the intended groups are present, and use an appropriate numerical summary to inspect the distributions.


```python
# Check for missing values and summarize distributions by day
display(citywide_daily.isnull().sum())
display(citywide_daily.groupby("day_of_week")["request_count"].describe())
```

**What did your checks establish?**

Your answer: The checks establish that there are zero missing values. Furthermore, the summary statistics prove that all 7 days of the week are present in the dataset, and minimum/maximum counts are safely within a realistic range for city-wide data.

## Step 5: Evidence

Create a histogram, box plot, or violin plot that directly addresses your question. Include a clear title and descriptive axis labels.


```python
sns.boxplot(
    data=citywide_daily,
    x="request_count",
    y="day_of_week",
    color="coral"
)
plt.title("NYC 311 Street-Condition Requests by Day of the Week (2025)")
plt.xlabel("Total daily requests")
plt.ylabel("Day of the week")
plt.show()
```

**What pattern in the figure provides the strongest evidence for your answer?**

Your answer: The median lines and the interquartile ranges for the weekdays (Monday–Friday) sit noticeably further to the right on the x-axis compared to Saturday and Sunday. The weekend distributions are visibly lower and tighter.

## Step 6: Conclusion


**State one conclusion that directly answers your research question.**

Your answer: The daily volume of street condition requests is significantly higher on weekdays compared to the weekend, indicating that New Yorkers are less likely to report 311 street issues on Saturdays and Sundays.

## Step 7: Limitation


**State one important limitation on how your result should be interpreted.**

Your answer: This result highlights human behavioral patterns rather than physical street conditions. Potholes do not disappear on weekends; rather, people are likely commuting less, taking a break from administrative tasks, or delaying their reporting until the workweek begins.
