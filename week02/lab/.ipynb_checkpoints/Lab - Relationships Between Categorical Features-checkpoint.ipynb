# Analysis 1: Is churn related to contract type?

Follow the analysis workflow from question through limitation.


**Before calculating a cross-tab, which contract type do you predict had the greatest churn percentage? Explain your prediction.**

Your answer: Month-to-month contracts likely had the greatest churn percentage. Customers on these contracts are not locked in and face no early termination fees, making it much easier for them to leave at any time.

## Step 1: Question


The question is: **Is churn related to contract type?**


## Step 2: Data

The relevant features are `contract` and `churn`. Before calculating percentages, compare the raw counts in their joint distribution.

Create and display `contract_churn_counts`, with contract types in the rows and churn outcomes in the columns.

Create and display `contract_churn_counts`, with contract types in the rows and churn outcomes in the columns.


```python
contract_churn_counts = pd.crosstab(customers['contract'], customers['churn'])
print(contract_churn_counts)
```

**Which contract type has the greatest raw number of customers who churned?**

Your answer: The Month-to-month contract type has the greatest raw number of customers who churned.

**Do the raw churn counts establish which contract group was most likely to churn? Explain.**

Your answer: No, raw counts do not account for the total number of customers in each group. A group might have a high raw churn count simply because it has a significantly larger overall customer base, not because those customers are inherently more likely to leave.

## Step 3: Operation

To compare likelihood fairly, calculate churn percentages within each contract type. First inspect the group totals.

Recreate the contract/churn cross-tab using `margins=True`.


```python
pd.crosstab(customers['contract'], customers['churn'], margins=True)
```

**Why do the contract-group totals matter when comparing likelihood?**

Your answer: The group totals provide the denominator needed to calculate percentages. We must divide the churn count by the total number of people in that specific contract group to fairly compare rates across groups of different sizes.

### Joint percentages

Create and display `contract_joint_percent`. All cells together should add to 100%.


```python
contract_joint_percent = pd.crosstab(customers['contract'], customers['churn'], normalize='all') * 100
print(contract_joint_percent)
```

## Step 4: Check

Display the sum of every value in `contract_joint_percent`.


```python
print(contract_joint_percent.sum().sum())
```

**What denominator does each joint percentage use?**

Your answer: Each joint percentage uses the total number of observations in the entire dataset.

**Does the joint-percentage table directly answer which contract group was most likely to churn?**

Your answer: No. It only shows what percentage of the entire company falls into each contract-and-churn bucket, rather than the isolated churn rate strictly within a specific contract group.

### Conditional percentages

Create `churn_within_contract` so each row describes churn outcomes within one contract type.


```python
churn_within_contract = pd.crosstab(customers['contract'], customers['churn'], normalize='index') * 100
print(churn_within_contract)
```

### Check the conditional percentages

Display the sum of each row in `churn_within_contract`.


```python
print(churn_within_contract.sum(axis=1))
```

**What should every row add to, and why?**

Your answer: Every row should add up to 100 (or 100%). This is because we normalized across the index (rows), forcing each contract group to act as its own independent population.

**Which contract group had the greatest observed churn percentage? Report the percentage.**

Your answer: The Month-to-month contract group had the greatest observed churn percentage at 42.71%.

**Was your prediction supported?**

Your answer: Yes, the prediction was supported.

## Step 5: Evidence

### Visualize the raw counts

Make a stacked bar chart from `contract_churn_counts`.


```python
contract_churn_counts.plot(kind='bar', stacked=True, figsize=(8, 5))
plt.title('Raw Counts of Churn by Contract Type')
plt.xlabel('Contract Type')
plt.ylabel('Number of Customers')
plt.xticks(rotation=0)
plt.legend(title='Churn')
plt.show()
```

### Visualize the conditional percentages

Make a 100% stacked bar chart from `churn_within_contract`. Use:

- `"lightgray"` for customers who did not churn;
- `"steelblue"` for customers who churned;
- a question-focused title;
- descriptive axis labels;
- horizontal x-axis labels; and
- an audience-friendly legend.


```python
ax = churn_within_contract.plot(
    kind='bar', 
    stacked=True, 
    color=['lightgray', 'steelblue'], 
    figsize=(8, 5)
)
plt.title('Percentage of Customer Churn by Contract Type')
plt.xlabel('Contract Type')
plt.ylabel('Percentage of Customers (%)')
plt.xticks(rotation=0)
plt.legend(title='Customer Outcome', labels=['Did Not Churn', 'Churned'], bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

**Which chart answers the likelihood question more clearly: counts or conditional percentages? Explain.**

Your answer: The conditional percentages chart answers it more clearly. By scaling every bar to equal 100%, it removes the distraction of unequal group sizes and makes it visually obvious that a much larger proportion of month-to-month customers churned compared to the other groups.

## Step 6: Conclusion


**What conclusion about contract type and churn is supported by this sample?**

Your answer: The sample supports the conclusion that customers on a month-to-month contract are much more likely to churn compared to customers on one-year or two-year contracts.

## Step 7: Limitation


**Why does this analysis not prove that changing a customer's contract would cause a change in churn?**

Your answer: This is observational data, which only shows correlation, not causation. Simply moving a month-to-month customer to a two-year contract wouldn't automatically prevent them from churning; the contract choice is likely influenced by lurking variables, such as a customer already planning to move or cancel soon.

# Analysis 2: Is churn related to payment method?

Apply the same workflow to a second pair of categorical features.


## Step 1: Question


The question is: **Is churn related to payment method?**


## Step 2: Data

Create and display the joint distribution of `payment_method` and `churn` as raw counts. Name it `payment_churn_counts`.

Create and display `payment_churn_counts`.


```python
payment_churn_counts = pd.crosstab(customers['payment_method'], customers['churn'])
print(payment_churn_counts)
```

## Step 3: Operation

Create and display `churn_within_payment`.


```python
churn_within_payment = pd.crosstab(customers['payment_method'], customers['churn'], normalize='index') * 100
print(churn_within_payment)
```

## Step 4: Check

Display the sum of every row in `churn_within_payment`.


```python
print(churn_within_payment.sum(axis=1))
```

**Which payment-method group had the greatest observed churn percentage? Report the percentage.**

Your answer: The Electronic check group had the greatest observed churn percentage at 45.29%.

## Step 5: Evidence

Make a 100% stacked bar chart. Choose two readable colors that are different from the `"lightgray"` and `"steelblue"` used in the contract chart. Use the same color consistently for each outcome throughout this chart and include audience-friendly labels.


```python
ax = churn_within_payment.plot(
    kind='bar', 
    stacked=True, 
    color=['#d3d3d3', '#ff9999'],
    figsize=(10, 5)
)
plt.title('Percentage of Customer Churn by Payment Method')
plt.xlabel('Payment Method')
plt.ylabel('Percentage of Customers (%)')
plt.xticks(rotation=0)
plt.legend(title='Customer Outcome', labels=['Did Not Churn', 'Churned'], bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

## Step 6: Conclusion


**What conclusion about payment method and churn is supported by the sample? Include the percentage that provides your strongest evidence.**

Your answer: The sample supports the conclusion that payment method is related to churn likelihood. Customers using Electronic checks experienced a churn rate of 45.29%, which is significantly higher than the rates observed for mailed checks, bank transfers, or credit cards.

## Step 7: Limitation


**Why does this analysis not prove that changing a customer's payment method would cause a change in churn?**

Your answer: This analysis does not prove that changing the payment method itself causes a change in churn. Payment method might just act as a proxy for other variables like demographic differences, technological comfort, or financial stability, which are the actual drivers of churn.
